```python
import pandas as pd

file_name = 'fulan.xlsx'
dataraw = pd.read_excel(file_name)

print("DATARAW:")
print(dataraw)
```

```python
datafrq = pd.crosstab(index=dataraw["Grade"], columns="Frekuensi")
print("Tabel Frekuensi:")
print(datafrq)
```

```python
import matplotlib.pyplot as plt

plt.figure()
datafrq.plot(kind='line', marker='o')
plt.title('Grafik Garis Distribusi Grade')
plt.xlabel('Grade')
plt.ylabel('Frekuensi')
plt.grid(True)
plt.show()
```

```python
plt.figure()
datafrq.plot(kind='bar', legend=False)
plt.title('Grafik Batang Distribusi Grade')
plt.xlabel('Grade')
plt.ylabel('Frekuensi')
plt.show()
```

```python
plt.figure()
datafrq.plot(kind='pie', y='Frekuensi', autopct='%1.0f%%', legend=False)
plt.title('Pie Chart Distribusi Grade')
plt.ylabel('')
plt.show()
```

```python
dataraw["Final Score"] = pd.to_numeric(dataraw["Final Score"], errors='coerce')
dt = dataraw["Final Score"]
stats = dt.describe()
stats['Standard Error'] = dt.sem()
stats['Median'] = dt.median()
stats['Mode'] = dt.mode().iloc[0]
stats['variance'] = dt.var()
stats['range'] = dt.max() - dt.min()
stats['skewness'] = dt.skew()
stats['kurtosis'] = dt.kurtosis()
print("STATISTIKA DESKRIPTIF NILAI: ")
print(stats)
```
